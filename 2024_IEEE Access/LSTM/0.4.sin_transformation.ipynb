```python
import pandas as pd
import os
import numpy as np
```

```python
#path = 'data/clean'
path = 'data/sampled/s15'
```

```python
ficheros = [f for f in os.listdir(path) if f.endswith('.parquet')]
```

```python
!mkdir data\sampled\s15\sector
#añado la columna sector
for fichero in ficheros:
    df = pd.read_parquet(os.path.join(path, fichero))
    bins = [0,180,360]
    labels = [0,1]
    df['sector'] = pd.cut(df['track'], bins = bins, labels = labels, right = False)
    
    nombre_nuevo_fichero = 'sector/' + fichero  
    ruta_nuevo_fichero = os.path.join(path, nombre_nuevo_fichero) 
    df.to_parquet(ruta_nuevo_fichero)
```

```python
!mkdir data\sampled\s15\sin
for fichero in ficheros:
    df = pd.read_parquet(os.path.join(path, fichero))
    
    df['track'] = np.deg2rad(df.track)
    df['track'] = df['track'].apply(lambda x:np.sin(x/2))

    nombre_nuevo_fichero = 'sin/' + fichero  
    ruta_nuevo_fichero = os.path.join(path, nombre_nuevo_fichero) 
    df.to_parquet(ruta_nuevo_fichero)
```
